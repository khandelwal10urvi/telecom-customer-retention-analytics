```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()
```

```python

import io

file_name = next(iter(uploaded))

df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("Dataset loaded successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python

print("Dataset dimensions:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()

print("\nFirst 5 records:")
display(df.head())

print("\nSummary of numerical columns:")
display(df.describe())
```

```python
# STEP 2: DATA QUALITY AUDIT

print("Missing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nUnique values in TotalCharges:")
print("TotalCharges dtype:", df["TotalCharges"].dtype)
print("Blank/empty values:", (df["TotalCharges"].astype(str).str.strip() == "").sum())

print("\nChurn distribution:")
display(df["Churn"].value_counts())

print("\nChurn percentage:")
display(df["Churn"].value_counts(normalize=True).mul(100).round(2))

print("\nUnique values in SeniorCitizen:")
print(df["SeniorCitizen"].unique())

print("\nUnique values in Contract:")
print(df["Contract"].unique())

print("\nUnique values in PaymentMethod:")
print(df["PaymentMethod"].unique())
```

```python
# STEP 3: INVESTIGATE BLANK TOTAL CHARGES

blank_total = df[df["TotalCharges"].astype(str).str.strip() == ""]

print("Number of blank TotalCharges records:", len(blank_total))

display(
    blank_total[
        [
            "customerID",
            "tenure",
            "MonthlyCharges",
            "Contract",
            "PaymentMethod",
            "Churn"
        ]
    ]
)
```

```python
# STEP 4: CLEAN TOTAL CHARGES

# Convert blank strings to NaN
df["TotalCharges"] = df["TotalCharges"].replace(r"^\s*$", np.nan, regex=True)

# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Fill the 11 blank TotalCharges values with 0
# Reason: all affected customers have tenure = 0
df.loc[df["TotalCharges"].isna() & (df["tenure"] == 0), "TotalCharges"] = 0

print("Remaining missing TotalCharges:", df["TotalCharges"].isna().sum())
print("TotalCharges dtype:", df["TotalCharges"].dtype)

display(df[["tenure", "MonthlyCharges", "TotalCharges"]].head())
```

```python
# STEP 5: VALIDATE TOTAL CHARGES

print("Customers with tenure = 0:")
display(
    df[df["tenure"] == 0][
        ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
    ]
)

print("\nTotal customers with tenure = 0:", (df["tenure"] == 0).sum())

print("\nTotalCharges summary:")
display(df["TotalCharges"].describe())
```

```python
# STEP 6: CREATE TENURE BAND

df["TenureBand"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, float("inf")],
    labels=["0-12 Months", "13-24 Months", "25-48 Months", "49+ Months"]
)

print("Tenure bands created successfully.")

display(
    df[["tenure", "TenureBand"]].head(10)
)

print("\nCustomers in each tenure band:")
display(df["TenureBand"].value_counts().sort_index())
```

```python
# STEP 7: CREATE BUSINESS-FRIENDLY FIELDS

# Churn flag: 1 = Churned, 0 = Retained
df["ChurnFlag"] = df["Churn"].map({"Yes": 1, "No": 0})

# Monthly charge bands
df["MonthlyChargeBand"] = pd.cut(
    df["MonthlyCharges"],
    bins=[-1, 30, 60, 90, float("inf")],
    labels=["Low (≤30)", "Medium (31-60)", "High (61-90)", "Very High (>90)"]
)

print("Business fields created successfully.")

display(
    df[
        [
            "MonthlyCharges",
            "MonthlyChargeBand",
            "Churn",
            "ChurnFlag"
        ]
    ].head(10)
)

print("\nChurnFlag distribution:")
display(df["ChurnFlag"].value_counts())

print("\nMonthly charge band distribution:")
display(df["MonthlyChargeBand"].value_counts().sort_index())
```

```python
# STEP 8: FINAL VALIDATION

print("FINAL DATASET CHECK")
print("-" * 40)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df[["TotalCharges", "tenure", "MonthlyCharges", "ChurnFlag"]].dtypes)

print("\nFinal columns:")
print(df.columns.tolist())
```

```python
# EXPORT CLEANED DATASET FOR EXCEL

output_file = "Telecom_Customer_Retention_Analytics.xlsx"

df.to_excel(output_file, index=False)

print(f"File exported successfully: {output_file}")
```
